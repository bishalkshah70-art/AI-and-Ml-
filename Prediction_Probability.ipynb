```python
import tensorflow as tf
from tensorflow.keras import layers,models
import numpy as np
```

```python
IMG_HEIGHT = 128
IMG_WIDTH = 128

CHANNELS = 3
BATCH_SIZE =32
```

```python
# 2. Construct CNN Architecture
model = models.Sequential([
# --- Block 1: Low-Level Feature Extraction ---
# 32 filters of 3x3 size. Input: (128, 128, 3) -> Output: (128, 128, 32)
layers.Conv2D(32, (3, 3), activation='relu', padding='same',
input_shape=(IMG_HEIGHT, IMG_WIDTH, CHANNELS)),
# Reduces spatial size by half -> Output: (64, 64, 32)
layers.MaxPooling2D((2, 2)),
# --- Block 2: Mid-Level Feature Extraction ---
# 64 filters look for combinations of shapes -> Output: (64, 64, 64)
layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
# Downsample -> Output: (32, 32, 64)
layers.MaxPooling2D((2, 2)),
# --- Block 3: High-Level Feature Extraction ---
# 128 filters look for complex features (eyes, ears, snouts) -> Output: (32, 32, 128)
layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
# Downsample -> Output: (16, 16, 128)
layers.MaxPooling2D((2, 2)),
# --- Block 4: Flattening & Classification ---
# Unroll (16, 16, 128) 3D matrix into a 1D array of size: 16 * 16 * 128 = 32,768elements
layers.Flatten(),
# Dense hidden layer with 128 neurons for feature combination
layers.Dense(128, activation='relu'),
# Dropout layer randomly turns off 50% of nodes during training to prevent overfitting
layers.Dropout(0.5),
# Output Layer: Single node with Sigmoid activation for Binary Classification
# Output <= 0.5 -> Cat | Output > 0.5 -> Dog
layers.Dense(1, activation='sigmoid')
])
```

```python
model.compile(
optimizer='adam', # Adaptive Learning Rate Optimizer
loss='binary_crossentropy', # Loss function for 2-class setup
metrics=['accuracy'] # Track evaluation metrics
)
model.summary()
```

```python

# Create dummy random image tensor representing a single loaded test sample
dummy_image = np.random.rand(1, IMG_HEIGHT, IMG_WIDTH, CHANNELS).astype(np.float32)

# Pass image through network
prediction_probability = model.predict(dummy_image)[0][0]

# Decision Thresholding
if prediction_probability > 0.5:
    label = "Dog"
    confidence = prediction_probability * 100
else:
    label = "Cat"
    confidence = (1 - prediction_probability) * 100

print(f"\nInference Prediction: {label} ({confidence:.2f}% Confidence)")
```
